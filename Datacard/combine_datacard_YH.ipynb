```python
import os
import glob

cat_name = ['cat34FHlowpurity', 'cat34FHhighpurity', 'cat34SLhighpurity', 'cat34SLlowpurity']
mass_name = [
    "MX400_MY250", "MX500_MY250", "MX500_MY300", "MX600_MY250", "MX600_MY300", "MX600_MY400",
    "MX700_MY250", "MX700_MY300", "MX700_MY400", "MX700_MY500",
    "MX800_MY250", "MX800_MY300", "MX800_MY400", "MX800_MY500", "MX800_MY600",
    "MX900_MY250", "MX900_MY300", "MX900_MY400", "MX900_MY500", "MX900_MY600", "MX900_MY700",
    "MX1000_MY250", "MX1000_MY300", "MX1000_MY400", "MX1000_MY500", "MX1000_MY600", "MX1000_MY700", "MX1000_MY800"]
mass_name = ["MX1000_MY400"]

year = '2018'
# year = '2016post'

# Output file — will be overwritten each time you run the script
output_file = f"run_combinecat34_{year}.sh"

with open(output_file, 'w', encoding='utf-8') as f:
    f.write(f"# Generated commands for year = {year}\n")
    f.write(f"# Date: {os.popen('date').read().strip()}\n\n")

    for mass in mass_name:
        no_cat = []
        f.write(f"\n# Mass point: {mass}\n")

        for cat in cat_name:
            mass_dir = "datacard_" + mass

            FHSLcat_path = f"/afs/cern.ch/user/s/shsong/CMSSW_10_6_20/src/flashggFinalFit/Datacard/Datacard_{mass}_{year}_wwgg_{cat}.txt"
            vbfcat_path  = f"/afs/cern.ch/user/s/shsong/CMSSW_10_6_20/src/flashggFinalFit/Datacard/Datacard_{mass}_{year}_VBF_{cat}.txt"
            tthcat_path  = f"/afs/cern.ch/user/s/shsong/CMSSW_10_6_20/src/flashggFinalFit/Datacard/Datacard_{mass}_{year}_TTH_{cat}.txt"
            vhcat_path   = f"/afs/cern.ch/user/s/shsong/CMSSW_10_6_20/src/flashggFinalFit/Datacard/Datacard_{mass}_{year}_VH_{cat}.txt"
            gghcat_path  = f"/afs/cern.ch/user/s/shsong/CMSSW_10_6_20/src/flashggFinalFit/Datacard/Datacard_{mass}_{year}_GGH_{cat}.txt"

            FHSLws_path = f"/eos/cms/store/group/phys_b2g/shsong/YH_flashggws/cat34/{year}/{mass}/ws_ggyhwwgg_{mass}_{year}_{cat}_wwgg"
            vbfws_path  = f"/eos/cms/store/group/phys_b2g/shsong/YH_flashggws/cat34/{year}/{mass}/ws_ggyhVBF_{mass}_{year}_{cat}_VBF"
            tthws_path  = f"/eos/cms/store/group/phys_b2g/shsong/YH_flashggws/cat34/{year}/{mass}/ws_ggyhTTH_{mass}_{year}_{cat}_TTH"
            vhws_path   = f"/eos/cms/store/group/phys_b2g/shsong/YH_flashggws/cat34/{year}/{mass}/ws_ggyhVH_{mass}_{year}_{cat}_VH"
            gghws_path  = f"/eos/cms/store/group/phys_b2g/shsong/YH_flashggws/cat34/{year}/{mass}/ws_ggyhGGH_{mass}_{year}_{cat}_GGH"

            FHSLsignalmodel = glob.glob(os.path.join(FHSLws_path, 'CMS-HGG_sigfit_packaged_*.root'))
            vbfsignalmodel  = glob.glob(os.path.join(vbfws_path,  'CMS-HGG_sigfit_packaged_*.root'))
            tthsignalmodel  = glob.glob(os.path.join(tthws_path,  'CMS-HGG_sigfit_packaged_*.root'))
            vhsignalmodel   = glob.glob(os.path.join(vhws_path,   'CMS-HGG_sigfit_packaged_*.root'))
            gghsignalmodel  = glob.glob(os.path.join(gghws_path,  'CMS-HGG_sigfit_packaged_*.root'))

            FHSLsignaltxt = os.path.exists(FHSLcat_path) and os.path.getsize(FHSLcat_path) > 1000
            vbfsignaltxt  = os.path.exists(vbfcat_path)  and os.path.getsize(vbfcat_path)  > 1000
            tthsignaltxt  = os.path.exists(tthcat_path)  and os.path.getsize(tthcat_path)  > 1000
            vhsignaltxt   = os.path.exists(vhcat_path)   and os.path.getsize(vhcat_path)   > 1000
            gghsignaltxt  = os.path.exists(gghcat_path)  and os.path.getsize(gghcat_path)  > 1000

            FHSL_has_model = len(FHSLsignalmodel) > 0 and FHSLsignaltxt
            vbf_has_model  = len(vbfsignalmodel)  > 0 and vbfsignaltxt
            tth_has_model  = len(tthsignalmodel)  > 0 and tthsignaltxt
            vh_has_model   = len(vhsignalmodel)   > 0 and vhsignaltxt
            ggh_has_model  = len(gghsignalmodel)  > 0 and gghsignaltxt
            if not FHSL_has_model:
                f.write(f"# SKIP {cat} — no FHSL model\n")
            else:
                if "16" in year:
                    yieldyear = "16"
                else:
                    yieldyear = year.split("20")[-1]

                ifprocess = [FHSL_has_model, vbf_has_model, tth_has_model, vh_has_model, ggh_has_model]
                process   = ["ggyhwwgg", "ggyhVBF", "ggyhTTH", "ggyhVH", "ggyhGGH"]

                existed_process = ""
                for i in range(len(ifprocess)):
                    if ifprocess[i]:
                        existed_process += process[i] + ","

                existed_process = existed_process.rstrip(",")
                process_list = existed_process.split(",")

                yieldcommand = (
                    f"python RunYields_{yieldyear}resolved_combined.py "
                    f"--cats {cat} "
                    f"--inputWSDirMap {year}=/eos/cms/store/group/phys_b2g/shsong/YH_flashggws/cat34/{year}/{mass}/ws_ggyhcombine_{mass}_{year}_{cat} "
                    f"--procs {existed_process} "
                    f"--sigModelWSDir {year}=/eos/cms/store/group/phys_b2g/shsong/YH_flashggws/cat34/{year}/{mass}/ws_ggyhcombine_{mass}_{year}_{cat} "
                    f"--bkgModelWSDir /eos/cms/store/group/phys_b2g/shsong/YH_flashggws/cat34/{year}/{mass}/ws_{year}_{cat} "
                    f"--cats {cat} --doSystematics --batch local "
                    f"--ext {mass}_{year}_{cat}"
                )

                if len(process_list) == 0:
                    no_cat.append(cat)
                    f.write(f"# SKIP {cat} — no valid processes\n")
                    continue

                elif len(process_list) == 1 and process_list[0] == "ggyhwwgg":
                    cpcommand = f"cp Datacard_{mass}_{year}_wwgg_{cat}.txt {mass}_{year}_{cat}.txt"
                    f.write(cpcommand + "\n")
                else:
                    # multi-process case
                    f.write(f"mkdir /eos/cms/store/group/phys_b2g/shsong/YH_flashggws/cat34/{year}/{mass}/ws_ggyhcombine_{mass}_{year}_{cat}\n")
                    f.write(f"cp /eos/cms/store/group/phys_b2g/shsong/YH_flashggws/cat34/{year}/{mass}/ws_ggyh*_{cat}*/*.root /eos/cms/store/group/phys_b2g/shsong/YH_flashggws/cat34/{year}/{mass}/ws_ggyhcombine_{mass}_{year}_{cat}/\n")
                    f.write(yieldcommand + "\n")

                    mkdatacard = (
                        f"python makeDatacard_{yieldyear}resolved_combined.py "
                        f"--years {year} --doSystematics --prune --ext {mass}_{year}_{cat} "
                        f"--output {mass}_{year}_{cat}"
                    )
                    f.write(mkdatacard + "\n")

                    f.write(f'strlist=($(grep "shapes *ggyh*" {mass}_{year}_{cat}.txt | awk \'{{print $2}}\' | cut -d\'_\' -f1))\n')
                    f.write('for str in "${strlist[@]}"; do\n')
                    f.write(f'    sed -i "0,/ws_ggyhcombine/s//ws_${{str}}/" {mass}_{year}_{cat}.txt\n')
                    f.write(f'    sed -i "0,/purity\\/CMS/s//purity_${{str#ggyh}}\\/CMS/" {mass}_{year}_{cat}.txt\n')
                    f.write("done\n")

                    f.write(f"sed -i 's/{year}=//g' {mass}_{year}_{cat}.txt\n")
                    f.write(f"rm -r /eos/cms/store/group/phys_b2g/shsong/YH_flashggws/cat34/{year}/{mass}/ws_ggyhcombine_{mass}_{year}_{cat}\n")
                    f.write(f"text2workspace.py {mass}_{year}_{cat}.txt -m 125\n")
                    f.write("\n")

        # After all categories for this mass
        combinecommand = (
            f"combineCards.py "
            f"cat34SLlowpurity_{year}={mass}_{year}_cat34SLlowpurity.txt "
            f"cat34SLhighpurity_{year}={mass}_{year}_cat34SLhighpurity.txt "
            f"cat34FHlowpurity_{year}={mass}_{year}_cat34FHlowpurity.txt "
            f"cat34FHhighpurity_{year}={mass}_{year}_cat34FHhighpurity.txt "
            f"> {mass}_{year}_cat34.txt"
        )

        if len(no_cat) > 0:
            if len(no_cat) == 3:
                only_cat = [c for c in cat_name if c not in no_cat]
                if only_cat:
                    combinecommand = f"cp {mass}_{year}_{only_cat[0]}.txt {mass}_{year}_cat34.txt"
            else:
                for cat in no_cat:
                    rm_part = f"{cat}_{year}={mass}_{year}_{cat}.txt"
                    combinecommand = combinecommand.replace(rm_part, "").strip()

        f.write(combinecommand + "\n")
        f.write(f"sed -i '/CMS_wwgg_br_HH_WWgg/d' {mass}_{year}_cat34.txt\n")
        f.write(f"text2workspace.py {mass}_{year}_cat34.txt -m 125\n")
        # f.write("\n" + "="*70 + "\n")

print(f"All commands have been written to: {output_file}")
```

```python
cat_name = ['cat34FHlowpurity', 'cat34FHhighpurity','cat34SLhighpurity','cat34SLlowpurity']
mass_name = ["MX400_MY250","MX500_MY250", "MX500_MY300", "MX600_MY250", "MX600_MY300", "MX600_MY400", "MX700_MY250", "MX700_MY300", "MX700_MY400", "MX700_MY500", "MX800_MY250", "MX800_MY300", "MX800_MY400", "MX800_MY500", "MX800_MY600", "MX900_MY250", "MX900_MY300", "MX900_MY400", "MX900_MY500", "MX900_MY600", "MX900_MY700","MX1000_MY250", "MX1000_MY300", "MX1000_MY400", "MX1000_MY500", "MX1000_MY600", "MX1000_MY700", "MX1000_MY800"]
# mass_name = ["MX800_MY400"]
# year='2017'
year='2016pre'
cd = []
for mass in mass_name:
    no_cat = []
    for cat in cat_name:
        mass_dir = "datacard_" + mass
        FHSLcat_path="/afs/cern.ch/user/s/shsong/CMSSW_10_6_20/src/flashggFinalFit/Datacard/Datacard_"+mass+"_"+year+"_wwgg_"+cat+".txt"
        vbfcat_path="/afs/cern.ch/user/s/shsong/CMSSW_10_6_20/src/flashggFinalFit/Datacard/Datacard_"+mass+"_"+year+"_VBF_"+cat+".txt"
        tthcat_path="/afs/cern.ch/user/s/shsong/CMSSW_10_6_20/src/flashggFinalFit/Datacard/Datacard_"+mass+"_"+year+"_TTH_"+cat+".txt"
        vhcat_path="/afs/cern.ch/user/s/shsong/CMSSW_10_6_20/src/flashggFinalFit/Datacard/Datacard_"+mass+"_"+year+"_VH_"+cat+".txt"
        gghcat_path="/afs/cern.ch/user/s/shsong/CMSSW_10_6_20/src/flashggFinalFit/Datacard/Datacard_"+mass+"_"+year+"_GGH_"+cat+".txt"
        FHSLws_path="/eos/cms/store/group/phys_b2g/shsong/YH_flashggws/cat34/"+year+"/"+mass+"/ws_ggyhwwgg_"+mass+"_"+year+"_"+cat+"_wwgg"
        vbfws_path="/eos/cms/store/group/phys_b2g/shsong/YH_flashggws/cat34/"+year+"/"+mass+"/ws_ggyhVBF_"+mass+"_"+year+"_"+cat+"_VBF"
        tthws_path="/eos/cms/store/group/phys_b2g/shsong/YH_flashggws/cat34/"+year+"/"+mass+"/ws_ggyhTTH_"+mass+"_"+year+"_"+cat+"_TTH"
        vhws_path="/eos/cms/store/group/phys_b2g/shsong/YH_flashggws/cat34/"+year+"/"+mass+"/ws_ggyhVH_"+mass+"_"+year+"_"+cat+"_VH"
        gghws_path="/eos/cms/store/group/phys_b2g/shsong/YH_flashggws/cat34/"+year+"/"+mass+"/ws_ggyhGGH_"+mass+"_"+year+"_"+cat+"_GGH"
        FHSLsignalmodel= glob.glob(os.path.join(FHSLws_path, 'CMS-HGG_sigfit_packaged_*.root'))
        vbfsignalmodel= glob.glob(os.path.join(vbfws_path, 'CMS-HGG_sigfit_packaged_*.root'))
        tthsignalmodel= glob.glob(os.path.join(tthws_path, 'CMS-HGG_sigfit_packaged_*.root'))
        vhsignalmodel= glob.glob(os.path.join(vhws_path, 'CMS-HGG_sigfit_packaged_*.root'))
        gghsignalmodel= glob.glob(os.path.join(gghws_path, 'CMS-HGG_sigfit_packaged_*.root'))
        FHSLsignaltxt = False
        vbfsignaltxt = False
        tthsignaltxt = False
        vhsignaltxt = False
        gghsignaltxt = False
        if os.path.exists(FHSLcat_path):
            FHSLsignaltxt = os.path.getsize(FHSLcat_path) > 1000
        if os.path.exists(vbfcat_path):
            vbfsignaltxt = os.path.getsize(vbfcat_path) > 1000
        if os.path.exists(tthcat_path):
            tthsignaltxt = os.path.getsize(tthcat_path) > 1000
        if os.path.exists(vhcat_path):
            vhsignaltxt = os.path.getsize(vhcat_path) > 1000
        if os.path.exists(gghcat_path):
            gghsignaltxt = os.path.getsize(gghcat_path) > 1000
        
        if len(FHSLsignalmodel)>0:
            if FHSLsignaltxt:
                FHSLsignalmodel = True
            else:
                FHSLsignalmodel = False
        else:
            FHSLsignalmodel = False        
        if len(vbfsignalmodel)>0:
            if vbfsignaltxt:
                vbfsignalmodel = True
            else:
                vbfsignalmodel = False
        else:
            vbfsignalmodel = False
        if len(tthsignalmodel)>0:
            if tthsignaltxt:
                tthsignalmodel = True
            else:
                tthsignalmodel = False
        else:
            tthsignalmodel = False
        if len(vhsignalmodel)>0:
            if vhsignaltxt:
                vhsignalmodel = True
            else:
                vhsignalmodel = False
        else:
            vhsignalmodel = False
            
        if len(gghsignalmodel)>0:
            if gghsignaltxt:
                gghsignalmodel = True
            else:
                gghsignalmodel = False
        else:
            gghsignalmodel = False
        if "16" in year:
            yieldyear="16"
        else:
            yieldyear=year.split("20")[-1]
        ifprocess=[FHSLsignalmodel, vbfsignalmodel, tthsignalmodel, vhsignalmodel, gghsignalmodel]
        process=["ggyhwwgg","ggyhVBF","ggyhTTH","ggyhVH","ggyhGGH"]
        existed_process =" "
        for i in range(len(ifprocess)):
            if ifprocess[i]:
                existed_process += (process[i])+","
        yieldcommand="python RunYields_"+yieldyear+"resolved_combined.py --cats "+cat+" --inputWSDirMap "+year+"=/eos/cms/store/group/phys_b2g/shsong/YH_flashggws/cat34/"+year+"/"+mass+"/ws_ggyhcombine_"+mass+"_"+year+"_"+cat+" --procs "+existed_process.replace(" ", "").rstrip(",")+" --sigModelWSDir "+year+"=/eos/cms/store/group/phys_b2g/shsong/YH_flashggws/cat34/"+year+"/"+mass+"/ws_ggyhcombine_"+mass+"_"+year+"_"+cat+" --bkgModelWSDir /eos/cms/store/group/phys_b2g/shsong/YH_flashggws/cat34/"+year+"/"+mass+"/ws_"+year+"_"+cat+" --cats "+cat+" --doSystematics --batch local --ext "+mass+"_"+year+"_"+cat+""
        process_list = existed_process.split(",")[:-1]
        if len(process_list) == 0:
            no_cat.append(cat)
            continue
        elif len(process_list) == 1 and " ggyhwwgg" in process_list:
            cpcommand="cp Datacard_"+mass+"_"+year+"_wwgg_"+cat+".txt "+mass+"_"+year+"_"+cat+".txt"
            print(cpcommand)
        elif len(process_list) == 1 and " ggyhwwgg" not in process_list:
            no_cat.append(cat)
            continue
        else:
            mkdircommand="mkdir /eos/cms/store/group/phys_b2g/shsong/YH_flashggws/cat34/"+year+"/"+mass+"/ws_ggyhcombine_"+mass+"_"+year+"_"+cat
            cpcommand="cp /eos/cms/store/group/phys_b2g/shsong/YH_flashggws/cat34/"+year+"/"+mass+"/ws_ggyh*"+cat+"*/*.root /eos/cms/store/group/phys_b2g/shsong/YH_flashggws/cat34/"+year+"/"+mass+"/ws_ggyhcombine_"+mass+"_"+year+"_"+cat+"/"
            replace4="sed -i 's/"+year+"=//g' "+mass+"_"+year+"_"+cat+".txt"
            mkdatacard="python makeDatacard_"+yieldyear+"resolved_combined.py --years "+year+" --doSystematics --prune --ext "+mass+"_"+year+"_"+cat+" --output "+mass+"_"+year+"_"+cat
            rmdircommand="rm -r /eos/cms/store/group/phys_b2g/shsong/YH_flashggws/cat34/"+year+"/"+mass+"/ws_ggyhcombine_"+mass+"_"+year+"_"+cat
            compilecommand="text2workspace.py "+mass+"_"+year+"_"+cat+".txt -m 125 "
            print(mkdircommand)
            print(cpcommand)
            print(yieldcommand)
            print(mkdatacard)
            process_txt='strlist=('+'$(grep "shapes *ggyh*" '+mass+'_'+year+'_'+cat+'.txt | awk \'{print $2}\' | cut -d\'_\' -f1))'
            print(process_txt)
            string = 'for str in "${strlist[@]}"; do'
            print(string)
            string = '    sed -i "0,/ws_ggyhcombine/s//ws_${str}/" \''+mass+'_'+year+'_'+cat+'.txt\''
            print(string)
            sedcommand='    sed -i "0,/purity\/CMS/s//purity_${str#ggyh}\/CMS/" \''+mass+'_'+year+'_'+cat+'.txt\''
            print(sedcommand)            
            print("done")
            print(replace4)
            print(rmdircommand)
            print(compilecommand)
            print(" ")
    combinecommand="combineCards.py  cat34SLlowpurity_"+year+"="+mass+"_"+year+"_cat34SLlowpurity.txt cat34SLhighpurity_"+year+"="+mass+"_"+year+"_cat34SLhighpurity.txt cat34FHlowpurity_"+year+"="+mass+"_"+year+"_cat34FHlowpurity.txt cat34FHhighpurity_"+year+"="+mass+"_"+year+"_cat34FHhighpurity.txt > "+mass+"_"+year+"_cat34.txt"
    if len(no_cat)>0 and len(no_cat)<3:
        for cat in no_cat:
            rmcat=cat+"_"+year+"="+mass+"_"+year+"_"+cat+".txt"
            #remove rmcat string from combinecommand
            combinecommand=combinecommand.replace(rmcat,"")
    elif len(no_cat)==3:
        only_cat = [cat for cat in cat_name if cat not in no_cat]
        combinecommand="cp "+mass+"_"+year+"_"+only_cat[0]+".txt "+mass+"_"+year+"_cat34.txt"
    rmcommand="sed -i '/CMS_wwgg_br_HH_WWgg/d' "+mass+"_"+year+"_cat34.txt"
    print(combinecommand)
    # print(ratecommand1)
    # print(ratecommand2)
    print(rmcommand)
    compilecommand="text2workspace.py "+mass+"_"+year+"_cat34.txt -m 125 "
    print(" ")
```
